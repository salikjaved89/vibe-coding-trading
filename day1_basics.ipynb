```python
try:
    a = float(input("Pehla number: "))
    b = float(input("Doosra number: "))
except ValueError:
    print("Galat input! Barah-e-meherbani sirf number likhein, jaise 5 ya 2.5")
else:
    op = input("Operation (+, -, *, /, %, **): ")

    if op == "+":
        print("Jawab:", a + b)
    elif op == "-":
        print("Jawab:", a - b)
    elif op == "*":
        print("Jawab:", a * b)
    elif op == "/":
        if b == 0:
            print("Zero se taqseem nahi ho sakti")
        else:
            print("Jawab:", a / b)
    elif op == "%":
        if b == 0:
            print("Zero se remainder nahi nikal sakte")
        else:
            print("Jawab:", a % b)
    elif op == "**":
        print("Jawab:", a ** b)
    else:
        print("Galat operation")
```

```python
try:
    a = float(input("Pehla number (trading mein entry price): "))
    b = float(input("Doosra number (trading mein exit price): "))
except ValueError:
    print("Galat input! Barah-e-meherbani sirf number likhein, jaise 5 ya 2.5")
else:
    op = input("Operation (+, -, *, /, %, **, p): ")

    if op == "+":
        print("Jawab:", a + b)
    elif op == "-":
        print("Jawab:", a - b)
    elif op == "*":
        print("Jawab:", a * b)
    elif op == "/":
        if b == 0:
            print("Zero se taqseem nahi ho sakti")
        else:
            print("Jawab:", a / b)
    elif op == "%":
        if b == 0:
            print("Zero se remainder nahi nikal sakte")
        else:
            print("Jawab:", a % b)
    elif op == "**":
        print("Jawab:", a ** b)
    elif op == "p":
        # Profit/Loss percentage = (exit - entry) / entry * 100
        if a == 0:
            print("Entry price zero ho to percentage nahi nikal sakte")
        else:
            nateeja = (b - a) / a * 100
            print("Profit/Loss:", round(nateeja, 2), "%")
    else:
        print("Galat operation")
```

```python
try:
    a = float(input("Pehla number (trading mein entry price): "))
    b = float(input("Doosra number (trading mein exit price): "))
except ValueError:
    print("Galat input! Barah-e-meherbani sirf number likhein, jaise 5 ya 2.5")
else:
    op = input("Operation (+, -, *, /, %, **, p): ")

    if op == "+":
        print("Jawab:", a + b)
    elif op == "-":
        print("Jawab:", a - b)
    elif op == "*":
        print("Jawab:", a * b)
    elif op == "/":
        if b == 0:
            print("Zero se taqseem nahi ho sakti")
        else:
            print("Jawab:", a / b)
    elif op == "%":
        if b == 0:
            print("Zero se remainder nahi nikal sakte")
        else:
            print("Jawab:", a % b)
    elif op == "**":
        print("Jawab:", a ** b)
    elif op == "p":
        # Profit/Loss percentage = (exit - entry) / entry * 100
        try:
            nateeja = (b - a) / a * 100
        except ZeroDivisionError:
            print("Entry price zero ho to percentage nahi nikal sakte")
        else:
            print("Profit/Loss:", round(nateeja, 2), "%")
    else:
        print("Galat operation")
```
